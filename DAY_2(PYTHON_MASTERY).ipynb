**Challenge 5 — Student Grade System**

```python
def student_grading_system(marks):
  if marks>=90:
    return "A"
  elif marks>=80:
    return "B"
  elif marks>=70:
    return "C"
  elif marks>=60:
    return "D"
  else:
    return "F"
student_grading_system(90)

```

**FINAL MINI-PROJECT DAY 2---Student Report Card**

```python
print("---STUDENT REPORT CARD---")
marks=[90,90,90,99,90,99]
def average_marks():
  average_marks1=sum(marks)/len(marks)
  return average_marks1
def grade():
  average_marks1=average_marks()
  if average_marks1>=90:
    return"A"
  elif average_marks1>=80:
    return"B"
  elif average_marks1>=70:
    return"C"
  elif average_marks1>=60:
    return"D"
  else:
    return"F"

def student_summary(name,city,age):
  print(f"Name:{name}")
  print(f"City:{city}")
  print(f"Age:{age}")
  print(f"Marks:{marks}")
  largest_marks=max(marks)
  smallest_marks=min(marks)
  print(f"The highest marks are {largest_marks}")
  print(f"The lowest marks are {smallest_marks}")
  if smallest_marks<70:
   print("Work hard next time")
  else:
   print("Excellent work")
print(average_marks())
print(grade())
student_summary("Amit","CHd",19)
```
