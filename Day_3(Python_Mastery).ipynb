```python
employees = [
    {"name": "Ani", "salary": 30000},
    {"name": "Amit", "salary": 35000},
    {"name": "Kenji", "salary": 40000}
]

for employee in employees:
  print(employee["name"],":",employee["salary"])
largest_salary = employees[0]["salary"]
print(largest_salary)
for employee in employees:
  if employee["salary"] > largest_salary:
    largest_salary = employee["salary"]
print(largest_salary)
```
