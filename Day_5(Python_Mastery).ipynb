**Challenge 6 — Find the Longest Word**

```python
cities = ("Delhi", "Chandigarh", "Mohali", "Mumbai","RAJGARH Shamti SOLAN")
for city in cities:
  print(city)
largest_city=cities[0]
for city in cities:
  if len(city)>len(largest_city):
    largest_city=city
print(f"The largest city={largest_city}")
```
