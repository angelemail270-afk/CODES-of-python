**Cybersecurity project**

```python
#Simple login system
allowed_users={"Kenji","Ani","Marry","John"}
user_name=input("Enter user_name:")
file=open("security_log2.txt","a")
if user_name in allowed_users:
  print("Access Granted")
  file.write(f"User={user_name}")
  file.write("\n Status: Login Success")
  file.close()
  file=open("security_log2.txt","r")
  content=file.read()
  print(content)
  file.close()
else:
  print("Access Denied")
  file.write(f"User={user_name}")
  file.write("\n Status: Login Failed")
  file.close()
  file=open("security_log2.txt","r")
  content=file.read()
  print(content)
  file.close()
  #Project success



```
